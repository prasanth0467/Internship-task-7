```python
import pymysql
try:
    conn=pymysql.connect(host="localhost",user="root",password="12345",database="sales")
    cursor=conn.cursor()
    print("success")
except Exception as e:
    print("Error",e) 
```

```python
import pandas as pd
query="""select market_code as market,
sum(sales_qty) as quantity,
sum(sales_amount) as revenue
from sales.transactions
group by market"""
df=pd.read_sql(query,conn)
print(df)
```

```python
import matplotlib.pyplot as plt
df.plot(kind="bar",x="market",y="revenue",legend=False)
plt.title("Revenue by market")
plt.ylabel("Revenue")
plt.xlabel("Market")
plt.tight_layout()
plt.savefig("Sales_chart.png")
plt.show()
```
